```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set a professional style and context
sns.set_theme(style="whitegrid", context="talk")

# Define your hyperparameters and metrics
hyperparams = ['alpha', 'beta', 'noise_sigma', 'noise_std', 'use_mean_z', 'detach']
metrics = ['obs_rmse', 'obs_nll', 'obs_elbo', 'obs_coverage', 'obs_uncertainty_corr', 'latent_ks',
           'miss_rmse', 'miss_nll', 'miss_elbo', 'miss_coverage', 'miss_uncertainty_corr', 'overall_coverage', 
           'training_time']

# Loop through each hyperparameter to create a separate figure
for param in hyperparams:
    # Create a figure with 2 rows and 6 columns (for the first 12 metrics)
    fig, axes = plt.subplots(2, 6, figsize=(24, 14), constrained_layout=True)
    fig.suptitle(f'Effect of {param} on Various Metrics', fontsize=24, y=1.02)
    
    # Loop over the first 12 metrics (exclude training_time if desired)
    for idx, metric in enumerate(metrics[:-1]):
        # Calculate subplot indices (row-major order: fill rows first)
        row = idx // 6
        col = idx % 6
        ax = axes[row, col]
        
        # Create a boxplot for the current metric vs. the hyperparameter
        sns.boxplot(data=df, x=param, y=metric, ax=ax, palette='viridis')
        
        # Set subplot title and labels with appropriate font sizes
        ax.set_title(metric, fontsize=18)
        ax.set_xlabel(param, fontsize=14)
        ax.set_ylabel(metric, fontsize=14)
        
        # Optionally, adjust tick parameters and remove the top/right spines
        ax.tick_params(axis='both', labelsize=12)
        sns.despine(ax=ax)
    
    plt.show()
```
