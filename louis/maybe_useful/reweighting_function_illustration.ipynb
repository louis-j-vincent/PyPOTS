```python
import numpy as np
from scipy.stats import norm
import matplotlib.pyplot as plt

xx = np.linspace(-5,5,200)

q_x_tilde = norm(0,1)
q_x = norm(-0.5, 0.7)

plt.plot(q_x.pdf(xx), label = "q_x");
plt.plot(q_x_tilde.pdf(xx), label = "q_x_corrupted");
plt.plot((q_x.pdf(xx) / q_x_tilde.pdf(xx)), label = "reweighting function : q_x/q_x_corrupted");
plt.plot((q_x.pdf(xx) / q_x_tilde.pdf(xx))**.5, label = "reweighting function : R(z,q_x,q_x_corrupted) for r = 0.5");
plt.legend(bbox_to_anchor=[0.5, 1.15], loc='center');
```
